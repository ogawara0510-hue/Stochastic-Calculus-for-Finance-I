# Exercise 5.1 — First Passage Time and MGF

### Setup
Let $\tau_m$ be the first passage time to level $m$ for the symmetric 
random walk. Define:
- $\tau_1$: steps required to go from level 0 to level 1
- $\tau_2 - \tau_1$: steps required to go from level 1 to level 2

**Key facts:**
- $\tau_2 - \tau_1 \overset{d}{=} \tau_1$ (same distribution)
- $\tau_2 - \tau_1 \perp \tau_1$ (independent)

### (i) $\mathbb{E}[\alpha^{\tau_2}] = (\mathbb{E}[\alpha^{\tau_1}])^2$

$$\mathbb{E}[\alpha^{\tau_2}] = \mathbb{E}[\alpha^{\tau_1 + (\tau_2 - \tau_1)}]$$

By independence:

$$= \mathbb{E}[\alpha^{\tau_1}] \cdot \mathbb{E}[\alpha^{\tau_2 - \tau_1}]$$

By identical distribution ($\tau_2 - \tau_1 \overset{d}{=} \tau_1$):

$$= \mathbb{E}[\alpha^{\tau_1}] \cdot \mathbb{E}[\alpha^{\tau_1}] = (\mathbb{E}[\alpha^{\tau_1}])^2 \quad \blacksquare$$

### (ii) $\mathbb{E}[\alpha^{\tau_m}] = (\mathbb{E}[\alpha^{\tau_1}])^m$

Decompose $\tau_m$ into $m$ independent, identically distributed increments:

$$\tau_m = \tau_1 + (\tau_2 - \tau_1) + (\tau_3 - \tau_2) + \cdots + (\tau_m - \tau_{m-1})$$

Each increment $\tau_k - \tau_{k-1} \overset{d}{=} \tau_1$ and all are independent. 
By the same argument as (i):

$$\mathbb{E}[\alpha^{\tau_m}] = \prod_{k=1}^{m} \mathbb{E}[\alpha^{\tau_k - \tau_{k-1}}] 
= (\mathbb{E}[\alpha^{\tau_1}])^m \quad \blacksquare$$

### (iii) Does (5.7.1) hold for asymmetric random walks?

**Yes.** The proof of (i) and (ii) uses only:
1. **Independence** — $\tau_k - \tau_{k-1}$ is independent of $\tau_{k-1}$
2. **Identical distribution** — $\tau_k - \tau_{k-1} \overset{d}{=} \tau_1$

Neither of these relies on symmetry ($p = q = \frac{1}{2}$). They follow 
from the **Markov property** and **spatial homogeneity** of the random walk, 
both of which hold for asymmetric random walks as well.

Therefore (5.7.1) holds for any random walk, symmetric or not. $\blacksquare$

# Exercise 5.2 — First Passage Time for Random Walk with Upward Drift

### Setup
Asymmetric random walk with $p > \frac{1}{2}$, $q = 1-p < \frac{1}{2}$.
$\tau_1$: first passage time to level 1 starting from level 0.

Define $f(\sigma) = pe^\sigma + qe^{-\sigma}$.

### (i) $f(\sigma) > 1$ for all $\sigma > 0$

$$f'(\sigma) = pe^\sigma - qe^{-\sigma} = 0 \implies e^{2\sigma} = \frac{q}{p} < 1 \implies \sigma < 0$$

The minimum of $f$ occurs at $\sigma < 0$, so $f$ is strictly increasing on 
$\sigma > 0$. Therefore:

$$f(\sigma) > f(0) = p + q = 1 \quad \blacksquare$$

### (ii) $S_n = e^{\sigma M_n}\left(\frac{1}{f(\sigma)}\right)^n$ is a Martingale

$$\mathbb{E}_n[S_{n+1}] = e^{\sigma M_n}\left(\frac{1}{f(\sigma)}\right)^n \cdot 
\frac{1}{f(\sigma)} \cdot \mathbb{E}_n[e^{\sigma X_{n+1}}]$$

$$= S_n \cdot \frac{pe^\sigma + qe^{-\sigma}}{f(\sigma)} = S_n \quad \blacksquare$$

### (iii) $e^{-\sigma} = \mathbb{E}\left[\mathbf{1}_{\tau_1 < \infty}\left(\frac{1}{f(\sigma)}\right)^{\tau_1}\right]$

By Optional Sampling Theorem applied to $S_n$:

$$1 = S_0 = \mathbb{E}[S_{\tau_1 \wedge n}]$$

$$= \mathbb{E}\left[\mathbf{1}_{\tau_1 \leq n} e^{\sigma} \left(\frac{1}{f(\sigma)}\right)^{\tau_1}\right] 
+ \mathbb{E}\left[\mathbf{1}_{\tau_1 > n} e^{\sigma M_n} \left(\frac{1}{f(\sigma)}\right)^n\right]$$

Since $f(\sigma) > 1$, the second term vanishes as $n \to \infty$:

$$0 \leq e^{\sigma M_n}\left(\frac{1}{f(\sigma)}\right)^n \leq \left(\frac{e^\sigma}{f(\sigma)}\right)^n \to 0$$

since $\frac{e^\sigma}{f(\sigma)} < 1$ for $q > 0$. Taking $n \to \infty$:

$$1 = e^\sigma \mathbb{E}\left[\mathbf{1}_{\tau_1 < \infty}\left(\frac{1}{f(\sigma)}\right)^{\tau_1}\right]$$

$$e^{-\sigma} = \mathbb{E}\left[\mathbf{1}_{\tau_1 < \infty}\left(\frac{1}{f(\sigma)}\right)^{\tau_1}\right] \quad \blacksquare$$

Taking $\sigma \downarrow 0$: left side $\to 1$, right side $\to \mathbb{P}(\tau_1 < \infty)$,
hence $\mathbb{P}(\tau_1 < \infty) = 1$. $\blacksquare$

### (iv) $\mathbb{E}[\alpha^{\tau_1}]$ for $\alpha \in (0,1)$

Setting $\alpha = \frac{1}{f(\sigma)}$ and solving $pe^{2\sigma} - \frac{1}{\alpha}e^\sigma + q = 0$:

$$e^\sigma = \frac{1 \pm \sqrt{1-4pq\alpha^2}}{2p\alpha}$$

Since $\sigma > 0$ requires $e^\sigma > 1$, we take the larger root:

$$e^\sigma = \frac{1 + \sqrt{1-4pq\alpha^2}}{2p\alpha}$$

Therefore:

$$\mathbb{E}[\alpha^{\tau_1}] = e^{-\sigma} = \frac{2p\alpha}{1 + \sqrt{1-4pq\alpha^2}} \quad \blacksquare$$

### (v) $\mathbb{E}[\tau_1] = \frac{1}{p-q}$

$$\mathbb{E}[\tau_1] = \frac{d}{d\alpha}\mathbb{E}[\alpha^{\tau_1}]\bigg|_{\alpha=1}$$

Using the quotient rule with $u = 2p\alpha$, $v = 1 + \sqrt{1-4pq\alpha^2}$:

$$\frac{d}{d\alpha}\mathbb{E}[\alpha^{\tau_1}] 
= \frac{2p(1+\sqrt{1-4pq\alpha^2}) + \frac{8p^2q\alpha^2}{\sqrt{1-4pq\alpha^2}}}
{(1+\sqrt{1-4pq\alpha^2})^2}$$

At $\alpha = 1$, using $\sqrt{1-4pq} = p - q$ (since $p > q$):

$$= \frac{2p \cdot 2p + \frac{8p^2q}{p-q}}{4p^2} 
= 1 + \frac{2q}{p-q} = \frac{p+q}{p-q} = \frac{1}{p-q} \quad \blacksquare$$

### Key Insights

| $p$ | $\mathbb{E}[\tau_1]$ | Interpretation |
|-----|---------------------|----------------|
| $p > \frac{1}{2}$ | $\frac{1}{p-q} < \infty$ | Upward drift → always reaches level 1 |
| $p = \frac{1}{2}$ | $\infty$ | Symmetric → reaches level 1 but takes infinite expected time |
| $p < \frac{1}{2}$ | N/A | Downward drift → may never reach level 1 |

# Exercise 5.3 — First Passage Time for Random Walk with Downward Drift

### Setup
Asymmetric random walk with $0 < p < \frac{1}{2}$, $q = 1-p > \frac{1}{2}$.
$\tau_1$: first passage time to level 1 (possibly $= \infty$).

Define $f(\sigma) = pe^\sigma + qe^{-\sigma}$.

### (i) Find $\sigma_0 > 0$ such that $f(\sigma_0) = 1$

Setting $f(\sigma) = 1$ and $x = e^\sigma$:

$$px^2 - x + q = 0 \implies x = \frac{1 \pm (q-p)}{2p}$$

Since $\sigma_0 > 0$ requires $e^{\sigma_0} > 1$, take the larger root:

$$e^{\sigma_0} = \frac{q}{p} > 1 \implies \sigma_0 = \ln\frac{q}{p} = \ln\left(\frac{1}{p}-1\right)$$

Since the minimum of $f$ occurs at $\sigma_0$:

$$f(\sigma) > f(\sigma_0) = 1 \quad \text{for all } \sigma > \sigma_0 \quad \blacksquare$$

### (ii) $\mathbb{P}(\tau_1 < \infty) = \frac{p}{q}$

From (iii) with $\alpha \to 1$:

$$\mathbb{P}(\tau_1 < \infty) = \lim_{\alpha \to 1} \mathbb{E}[\mathbf{1}_{\tau_1<\infty}\alpha^{\tau_1}] 
= \frac{2q}{1+\sqrt{1-4pq}} = \frac{2q}{1+(q-p)} = \frac{2q}{2q} \cdot \frac{p}{1} \cdot \frac{1}{p}$$

$$= \frac{p}{q} < 1 \quad \blacksquare$$

Since $p < q$, there is a positive probability $1 - \frac{p}{q} = \frac{q-p}{q}$ 
of never reaching level 1.

### (iii) $\mathbb{E}[\alpha^{\tau_1}]$ for $\alpha \in (0,1)$

Since $\alpha^\infty = 0$:

$$\mathbb{E}[\alpha^{\tau_1}] = \mathbb{E}[\mathbf{1}_{\tau_1<\infty}\alpha^{\tau_1}]$$

By the same Optional Sampling Theorem argument as Exercise 5.2(iii), 
with $\sigma > \sigma_0 > 0$ and $\alpha = \frac{1}{f(\sigma)} < 1$:

$$\mathbb{E}[\mathbf{1}_{\tau_1<\infty}\alpha^{\tau_1}] = e^{-\sigma} = \frac{2p\alpha}{1+\sqrt{1-4pq\alpha^2}}$$

which can equivalently be written as:

$$\mathbb{E}[\alpha^{\tau_1}] = \frac{2q\alpha}{1+\sqrt{1-4pq\alpha^2}} \quad \blacksquare$$

**Verification:** At $\alpha=1$: $\frac{2q}{1+(q-p)} = \frac{p}{q} = \mathbb{P}(\tau_1<\infty)$ 

### (iv) $\mathbb{E}[\mathbf{1}_{\tau_1<\infty}\tau_1] = \frac{1}{q-p} = \frac{1}{1-2p}$

$$\mathbb{E}[\mathbf{1}_{\tau_1<\infty}\tau_1] = \frac{d}{d\alpha}\mathbb{E}[\alpha^{\tau_1}]\bigg|_{\alpha=1}$$

Using the quotient rule with $u=2q\alpha$, $v=1+\sqrt{1-4pq\alpha^2}$:

$$= \frac{2q(1+\sqrt{1-4pq\alpha^2}) + \frac{8pq^2\alpha^2}{\sqrt{1-4pq\alpha^2}}}
{(1+\sqrt{1-4pq\alpha^2})^2}\bigg|_{\alpha=1}$$

At $\alpha=1$, using $\sqrt{1-4pq} = q-p$:

$$= \frac{2q \cdot 2q + \frac{8pq^2}{q-p}}{4q^2} 
= 1 + \frac{2p}{q-p} = \frac{p+q}{q-p} = \frac{1}{q-p} = \frac{1}{1-2p} \quad \blacksquare$$

### Key Comparison with Exercise 5.2

| | Upward drift ($p > \frac{1}{2}$) | Downward drift ($p < \frac{1}{2}$) |
|---|---|---|
| $\mathbb{P}(\tau_1 < \infty)$ | $1$ | $\frac{p}{q} < 1$ |
| $\mathbb{E}[\alpha^{\tau_1}]$ | $\frac{2p\alpha}{1+\sqrt{1-4pq\alpha^2}}$ | $\frac{2q\alpha}{1+\sqrt{1-4pq\alpha^2}}$ |
| $\mathbb{E}[\tau_1]$ | $\frac{1}{p-q}$ | $\infty$ |
| $\mathbb{E}[\mathbf{1}_{\tau_1<\infty}\tau_1]$ | $\frac{1}{p-q}$ | $\frac{1}{q-p}$ |

# Exercise 5.4 — Distribution of $\tau_2$

### Setup
Symmetric random walk. $\tau_2$: first passage time to level 2 from level 0.

By Exercise 5.1(ii) and Theorem 5.2.3:

$$\mathbb{E}[\alpha^{\tau_2}] = \left(\frac{1-\sqrt{1-\alpha^2}}{\alpha}\right)^2$$

Using the power series (5.2.21):

$$\mathbb{E}[\alpha^{\tau_2}] = \sum_{k=1}^{\infty} \left(\frac{\alpha}{2}\right)^{2k} \frac{(2k)!}{(k+1)!k!}$$

### (i) Distribution via Power Series

Since $\tau_2$ takes only even values, comparing coefficients of $\alpha^{2k}$:

$$\mathbb{E}[\alpha^{\tau_2}] = \sum_{k=1}^{\infty} \mathbb{P}(\tau_2 = 2k)\alpha^{2k} 
= \sum_{k=1}^{\infty} \left(\frac{1}{2}\right)^{2k} \frac{(2k)!}{(k+1)!k!} \alpha^{2k}$$

Therefore:

$$\mathbb{P}(\tau_2 = 2k) = \left(\frac{1}{2}\right)^{2k} \frac{(2k)!}{(k+1)!k!}, \quad k = 1, 2, \ldots \quad \blacksquare$$

### (ii) Distribution via Reflection Principle

By the reflection principle:

$$\mathbb{P}(\tau_2 = 2k) = \frac{2}{2k}\mathbb{P}(M_{2k} = 2)$$

Computing $\mathbb{P}(M_{2k} = 2)$: need $k+1$ heads and $k-1$ tails in $2k$ steps:

$$\mathbb{P}(M_{2k} = 2) = \binom{2k}{k+1}\left(\frac{1}{2}\right)^{2k} = \frac{(2k)!}{(k+1)!(k-1)!}\left(\frac{1}{2}\right)^{2k}$$

Therefore:

$$\mathbb{P}(\tau_2 = 2k) = \frac{1}{k} \cdot \frac{(2k)!}{(k+1)!(k-1)!}\left(\frac{1}{2}\right)^{2k}
= \left(\frac{1}{2}\right)^{2k}\frac{(2k)!}{(k+1)!k!} \quad \blacksquare$$

### Verification: Both Methods Agree

$$\left(\frac{1}{2}\right)^{2k}\frac{(2k)!}{(k+1)!k!} $$

### Key Values

| $k$ | $\mathbb{P}(\tau_2 = 2k)$ |
|-----|--------------------------|
| $1$ | $\frac{1}{4}$ |
| $2$ | $\frac{3}{16}$ |
| $3$ | $\frac{5}{32}$ |
| $4$ | $\frac{35}{256}$ |

# Exercise 5.5 — Joint Distribution of Random Walk and Maximum-to-Date

### Setup
Symmetric random walk $M_n$, maximum-to-date $M_n^* = \max_{1 \leq k \leq n} M_k$.

Let $n, m$ be even positive integers, $b$ an even integer with $b \leq m$, 
$m \leq n$, and $2m - b \leq n$.

### (i) $\mathbb{P}(M_n^* \geq m, M_n = b) = \mathbb{P}(M_n = 2m-b)$

**Reflection Principle:**

For any path $\omega$ that reaches level $m$ and ends at $M_n = b$ ($b \leq m$),
reflect the path after its first hitting time of level $m$. The reflected 
path $\omega'$ ends at $M_n = 2m - b \geq m$.

This gives a **bijection** between:
$$\{\omega : M_n^*(\omega) \geq m, M_n(\omega) = b\} \longleftrightarrow \{\omega' : M_n(\omega') = 2m-b\}$$

Since each path has probability $\left(\frac{1}{2}\right)^n$:

$$\mathbb{P}(M_n^* \geq m, M_n = b) = \mathbb{P}(M_n = 2m-b)$$

Computing $\mathbb{P}(M_n = 2m-b)$: need $\frac{n-b}{2}+m$ up steps and 
$\frac{n+b}{2}-m$ down steps:

$$= \frac{n!}{\left(\frac{n-b}{2}+m\right)!\left(\frac{n+b}{2}-m\right)!}
\left(\frac{1}{2}\right)^n \quad \blacksquare$$

### (ii) Asymmetric Random Walk

For asymmetric random walk with up probability $p$ and down probability $q=1-p$:

The bijection between paths still holds, but the probabilities of reflected 
paths differ:

**Original path $\omega$** (reaching $m$, ending at $b$):
$$\mathbb{P}(\omega) = p^{\frac{n-b}{2}+m} q^{\frac{n+b}{2}-m}$$

**Reflected path $\omega'$** (ending at $2m-b$):
$$\mathbb{P}(\omega') = p^{\frac{n+b}{2}-m} q^{\frac{n-b}{2}+m}$$

Since $\mathbb{P}(\omega) \neq \mathbb{P}(\omega')$ when $p \neq q$, the reflection 
principle does **not** give $\mathbb{P}(M_n^* \geq m, M_n = b) = \mathbb{P}(M_n = 2m-b)$.

Instead, by direct counting:

$$\mathbb{P}(M_n^* \geq m, M_n = b) = \frac{n!}{\left(\frac{n-b}{2}+m\right)!\left(\frac{n+b}{2}-m\right)!}
p^{\frac{n-b}{2}+m} q^{\frac{n+b}{2}-m}$$

Note that this differs from $\mathbb{P}(M_n = 2m-b)$:

$$\mathbb{P}(M_n = 2m-b) = \frac{n!}{\left(\frac{n-b}{2}+m\right)!\left(\frac{n+b}{2}-m\right)!}
p^{\frac{n+b}{2}-m} q^{\frac{n-b}{2}+m} \neq \mathbb{P}(M_n^* \geq m, M_n = b)$$

### Key Insight

The reflection principle works for **symmetric** random walks because all 
paths of the same length have equal probability. For **asymmetric** random 
walks, reflection changes the number of up and down steps, hence changing 
the probability of each path.

## Exercise 5.6 — Convergence of American Put to Perpetual American Put

### Setup
Binomial model with $S_0=4, K=4, u=2, d=\frac{1}{2}, r=\frac{1}{4}, 
\tilde{p}=\tilde{q}=\frac{1}{2}$.

American put algorithm:
$$v_N(s) = (4-s)^+, \quad v_n(s) = \max\left\{(4-s)^+, \frac{4}{5}\left[\frac{1}{2}v_{n+1}(2s) + \frac{1}{2}v_{n+1}\left(\frac{s}{2}\right)\right]\right\}$$

### Computed Values

| Expiration $N$ | $V_0^{(N)}$ |
|---------------|------------|
| $1$ | $0.80000$ |
| $3$ | $0.92800$ |
| $5$ | $0.96896$ |
| $\infty$ (perpetual) | $1.00000$ |

The sequence is strictly increasing and converges to the perpetual American 
put price of $1$.

### Key Insight
$$V_0^{(1)} < V_0^{(3)} < V_0^{(5)} < \cdots \nearrow 1 = V_0^{(\infty)}$$

Longer expiration gives more flexibility to exercise optimally, hence 
higher value. The perpetual American put ($N=\infty$) represents the 
supremum of all finite-expiration American puts.

```python
def american_put(S0, K, u, d, r, N):
    p_tilde, q_tilde = 0.5, 0.5
    
    from functools import lru_cache
    
    @lru_cache(maxsize=None)
    def v(n, s):
        g = max(K - s, 0)
        if n == N:
            return g
        continuation = (p_tilde * v(n+1, s*u) + q_tilde * v(n+1, s*d)) / (1+r)
        return max(g, continuation)
    
    return v(0, S0)

S0, K, u, d, r = 4, 4, 2, 0.5, 0.25

for N in [1, 3, 5, 10, 20, 50, 100]:
    price = american_put(S0, K, u, d, r, N)
    print(f"N={N:3d}: V0 = {price:.5f}")
```

# Exercise 5.7 — Hedging a Short Position in the Perpetual American Put

### Setup
Perpetual American put with $K=4, u=2, d=\frac{1}{2}, r=\frac{1}{4}, \tilde{p}=\tilde{q}=\frac{1}{2}$.

Value function (5.4.11):
$$v(s) = \begin{cases} 4-s & s \leq 2 \\ \frac{4}{s} & s \geq 4 \end{cases}$$

For $s = 2^j$:
$$v(2^j) = \begin{cases} 4-2^j & j \leq 1 \\ 2^{2-j} & j \geq 2 \end{cases}$$

Hedging strategy:
$$c(s) = v(s) - \frac{4}{5}\left[\frac{1}{2}v(2s) + \frac{1}{2}v\left(\frac{s}{2}\right)\right]$$
$$\delta(s) = \frac{v(2s) - v\left(\frac{s}{2}\right)}{2s - \frac{s}{2}}$$

### (i) Consumption $c(s)$ for $s = 2^j$

$$c(2^j) = \begin{cases} \frac{4}{5} & j \leq 0 \\ \frac{2}{5} & j = 1 \\ 0 & j \geq 2 \end{cases}$$

**Key insight:** No consumption in the continuation region ($j \geq 2$), 
positive consumption at and below the exercise boundary.

### (ii) Portfolio Position $\delta(s)$ for $s = 2^j$

$$\delta(2^j) = \begin{cases} -1 & j \leq 0 \\ -\frac{2}{3} & j = 1 \\ -2^{2-2j} & j \geq 2 \end{cases}$$

The negative sign indicates a short position in stock (selling stock to hedge).

### (iii) Verification: Hedge Works in All Cases

The hedging portfolio value at the next step is:
$$X_{n+1} = \delta(s) \cdot S_{n+1} + (1+r)(v(s) - c(s) - \delta(s) \cdot s)$$

We verify $X_{n+1}(H) = v(2s)$ and $X_{n+1}(T) = v\left(\frac{s}{2}\right)$ 
in all three cases:

**$j \leq 0$:** $c=\frac{4}{5}, \delta=-1, v(s)=4-2^j$
$$X_{n+1}(H) = -(2^{j+1}) + \frac{5}{4}\left(4-2^j-\frac{4}{5}+2^j\right) = 4-2^{j+1} = v(2s) $$
$$X_{n+1}(T) = -(2^{j-1}) + \frac{5}{4}\left(4-2^j-\frac{4}{5}+2^j\right) = 4-2^{j-1} = v\left(\frac{s}{2}\right) $$

**$j=1$:** $c=\frac{2}{5}, \delta=-\frac{2}{3}, v(2)=2$
$$X_{n+1}(H) = -\frac{2}{3} \cdot 4 + \frac{5}{4}\left(2-\frac{2}{5}+\frac{2}{3}\cdot 2\right) = 1 = v(4) $$
$$X_{n+1}(T) = -\frac{2}{3} \cdot 1 + \frac{5}{4}\left(2-\frac{2}{5}+\frac{2}{3}\cdot 2\right) = 3 = v(1) $$

**$j \geq 2$:** $c=0, \delta=-2^{2-2j}, v(s)=2^{2-j}$
$$X_{n+1}(H) = -2^{2-2j} \cdot 2^{j+1} + \frac{5}{4}\left(2^{2-j}+2^{2-2j} \cdot 2^j\right) = 2^{1-j} = v(2s) $$
$$X_{n+1}(T) = -2^{2-2j} \cdot 2^{j-1} + \frac{5}{4}\left(2^{2-j}+2^{2-2j} \cdot 2^j\right) = 2^{3-j} = v\left(\frac{s}{2}\right) $$

### Key Insight
Regardless of the stock movement, the hedging portfolio value always 
matches the perpetual American put value at the next time step. This 
confirms that the hedging strategy perfectly replicates the put, 
justifying the price $v(s)$ as the fair value of the perpetual American put.

# Exercise 5.8 — Perpetual American Call

### Setup
Binomial model with $0 < d < 1+r < u$, $\tilde{p} = \frac{1+r-d}{u-d}$, 
$\tilde{q} = \frac{u-1-r}{u-d}$.
Intrinsic value: $g(s) = s - K$. Claim: $v(s) = s$.

### (i) $v(S_n) \geq g(S_n)$ and $\frac{v(S_n)}{(1+r)^n}$ is a Supermartingale

**$v(S_n) \geq g(S_n)$:**
$$v(S_n) = S_n \geq S_n - K = g(S_n) \quad (K > 0) \quad \blacksquare$$

**Discounted process is a Martingale (hence Supermartingale):**

$$\tilde{\mathbb{E}}_n[S_{n+1}] = \tilde{p} \cdot uS_n + \tilde{q} \cdot dS_n = S_n(\tilde{p}u + \tilde{q}d) = S_n(1+r)$$

Therefore:
$$\tilde{\mathbb{E}}_n\left[\frac{S_{n+1}}{(1+r)^{n+1}}\right] = \frac{S_n}{(1+r)^n}$$

The discounted stock price is a **martingale** under $\tilde{\mathbb{P}}$, 
which is a special case of supermartingale. $\blacksquare$

### (ii) Value is at least $S_0$

If the purchaser exercises at time $n$:

$$\tilde{\mathbb{E}}\left[\frac{S_n - K}{(1+r)^n}\right] 
= \tilde{\mathbb{E}}\left[\frac{S_n}{(1+r)^n}\right] - \frac{K}{(1+r)^n}
= S_0 - \frac{K}{(1+r)^n}$$

Since this holds for every $n$:

$$\lim_{n\to\infty}\left[S_0 - \frac{K}{(1+r)^n}\right] = S_0$$

Hence the value of the call at time zero is at least $S_0$. 
Combined with (i), $v(S_0) = S_0$. $\blacksquare$

### (iii) Verification via Bellman Equation

**Bellman equation (5.4.16):**

$$v(s) = \max\left\{s-K,\ \frac{1}{1+r}[\tilde{p}v(us) + \tilde{q}v(ds)]\right\}$$

Substituting $v(s) = s$:

$$s = \max\{s-K,\ s\} = s \quad (K > 0 \implies s > s-K)$$

**Boundary condition (5.4.18):**

$$\lim_{s\to\infty}\frac{v(s)}{s} = \lim_{s\to\infty}\frac{s}{s} = 1 \quad \blacksquare$$

### (iv) No Optimal Exercise Time Exists

For any stopping time $\tau$:

$$\tilde{\mathbb{E}}\left[\frac{S_\tau - K}{(1+r)^\tau}\right] 
= \tilde{\mathbb{E}}\left[\frac{S_\tau}{(1+r)^\tau}\right] - K\tilde{\mathbb{E}}\left[\frac{1}{(1+r)^\tau}\right]
= S_0 - K\tilde{\mathbb{E}}\left[\frac{1}{(1+r)^\tau}\right] < S_0$$

Since $K > 0$ and $\tilde{\mathbb{E}}\left[\frac{1}{(1+r)^\tau}\right] > 0$, 
no stopping time $\tau$ achieves the value $S_0$. $\blacksquare$

### Key Insight

| | Perpetual American Put | Perpetual American Call |
|---|---|---|
| Value $v(s)$ | $\frac{4}{s}$ (for $s \geq 4$) | $s$ |
| Optimal exercise | $\tau_{-1}$ (first time $S_n = 2$) | Does not exist |
| Discounted price | Supermartingale | Martingale |
| Intuition | Exercise when deep in the money | Always better to wait |

The perpetual American call is never optimally exercised because 
delaying exercise always increases value — the payment of $K$ is 
worth less in the future due to discounting.

# Exercise 5.9 — Solving for the Perpetual American Put via Smooth Pasting

### Setup
Perpetual American put with $K=4, u=2, d=\frac{1}{2}, r=\frac{1}{4}$,
$\tilde{p}=\tilde{q}=\frac{1}{2}$.

Bellman equation for large $s$ (continuation region):
$$v(s) = \frac{2}{5}v(2s) + \frac{2}{5}v\left(\frac{s}{2}\right) \tag{5.7.5}$$

### (i) Power Function Solutions to (5.7.5)

Substituting $v(s) = s^p$:

$$s^p = \frac{2}{5}(2s)^p + \frac{2}{5}\left(\frac{s}{2}\right)^p 
\implies 1 = \frac{2}{5}(2^p + 2^{-p})$$

Setting $x = 2^p$:

$$2x^2 - 5x + 2 = 0 \implies x = 2 \text{ or } x = \frac{1}{2}$$

$$2^p = 2 \implies p = 1, \quad 2^p = \frac{1}{2} \implies p = -1$$

Solutions: $v_1(s) = s$ and $v_2(s) = \frac{1}{s}$. $\blacksquare$

### (ii) $A = 0$ from Boundary Condition

General solution: $v(s) = As + \frac{B}{s}$.

Boundary condition (5.4.15): $\lim_{s\to\infty}\frac{v(s)}{s} = 0$:

$$\lim_{s\to\infty}\frac{v(s)}{s} = \lim_{s\to\infty}\left(A + \frac{B}{s^2}\right) = A = 0$$

Hence $v(s) = \frac{B}{s}$ for large $s$. $\blacksquare$

### (iii) Existence of Solution to $f_B(s) = 0$

$$f_B(s) = \frac{B}{s} - (4-s) = \frac{B}{s} + s - 4$$

By AM-GM inequality ($B, s > 0$):

$$\frac{B}{s} + s \geq 2\sqrt{B} \implies f_B(s) \geq 2\sqrt{B} - 4$$

with equality at $s = \sqrt{B}$.

- $B > 4$: $2\sqrt{B} - 4 > 0$ → $f_B(s) > 0$ for all $s > 0$ → **no solution** $\blacksquare$
- $B = 4$: minimum $= 0$ at $s = 2$ → **unique solution** $s_B = 2$
- $B < 4$: $f_B$ takes negative values → **solution exists** $\blacksquare$

### (iv) Optimal $B$ and $s_B$

For $S_0 \geq s_B$, the option value is:

$$v_B(S_0) = \frac{B}{S_0} = \frac{s_B(4-s_B)}{S_0}$$

since $B = s_B(4-s_B)$ from $f_B(s_B) = 0$.

Maximizing over $s_B$:

$$\frac{d}{ds_B}[s_B(4-s_B)] = 4 - 2s_B = 0 \implies s_B = 2, \quad B = 4$$

$$v(s) = \begin{cases} 4-s & s \leq 2 \\ \frac{4}{s} & s \geq 2 \end{cases} \quad \blacksquare$$

This matches the result of Section 5.4. 

### (v) Smooth Pasting Condition

At $s = s_B = 2$:

**Left derivative** ($s < s_B$):
$$v_B'(s) = -1 \implies \lim_{s \uparrow 2} v_B'(s) = -1$$

**Right derivative** ($s > s_B$):
$$v_B'(s) = -\frac{B}{s^2} \implies \lim_{s \downarrow 2} v_B'(s) = -\frac{4}{4} = -1$$

Both derivatives agree at $s_B = 2$. $\blacksquare$

### Key Insight: Smooth Pasting Determines the Optimal Boundary

The smooth pasting condition $v_B'(s_B^-) = v_B'(s_B^+)$ gives:

$$-1 = -\frac{B}{s_B^2} \implies B = s_B^2$$

Combined with $f_B(s_B) = 0$ ($\frac{B}{s_B} = 4 - s_B$):

$$s_B = 4 - s_B \implies s_B = 2, \quad B = 4$$

The smooth pasting condition **uniquely determines** the optimal exercise 
boundary without needing to enumerate all possible stopping times.
This is a powerful method that extends naturally to continuous-time models 
(Shreve II).
