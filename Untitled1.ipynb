# Definition of Brownian Motion

A Browninan motion (or Wiener process) is a stochastic process

$$ {W(t)}_{t \geq 0} $$

that satisfies the following four properties:

1. **Initial condition**
   $$ W(0)=0$$
2. **Continuity**

   The sample paths of $W(t)$ are continuous with probability $1$.
   
3. **Independent increments**
   
   For any integer $n \geq 1$ and any times $0 \leq t_0 < t_1 < \cdots < t_n \leq T$,
   the random variables
   $$W(t_1)-W(t_0), W(t_2)-W(t_1), \ldots, W(t_n)-W(t_{n-1})$$
   are independent
   
4. **Normal increments**
   
   For all $0\leq s \leq t \leq T$, the increment $W(t)-W(s)$ follows a normal distribution with mean $0$ and varianve $(t-s)$:
   $$W(t)-W(s) \sim N(0, t-s)$$

   Especially, for s = 0, we abtain this from (i);
   $$W(t) \sim N(0, t)$$

```python
import numpy as np
import matplotlib.pyplot as plt

def brownian_motion(T,n):

    """
    Simulate one Brownian motion path.

    Parameters
    ----------
    T : float 
        Terminal time

    n: int
        number of time step

    Returns
    ----------
    t : ndarray
        Time grid

    W ; ndarray
        Brownian motion path
    """

    dt = T/n
    
    # time grid
    t = np.linspace(0, T, n+1)

    # Browninan increments
    dW = np.sqrt(dt) * np.random.randn(n)

    # Brownian path
    W = np.zeros(n+1)
    W[1:] = np.cumsum(dW)
    return t, W

t, W = brownian_motion(T=1, n=1000)
plt.plot(t, W)

plt.xlabel("t")
plt.ylabel("r$W_t$")
plt.title("Brownian Motion Path")

plt.grid(True)

plt.show()
```

```python
t, W = brownian_motion(T = 1, n = 1000)

plt.figure(figsize=(10,6))

for i in range(100):

    t, W = brownian_motion(T=1, n=1000)

    plt.plot(t, W, linewidth=0.8)

plt.xlabel("t")
plt.ylabel(r"$W_t$")
plt.title("100 Simulated Brownian Motion Paths")

plt.grid(True)

plt.show()
```

```python

# parameters
T = 1
n = 1000
num_paths = 100

# time grid
t = np.linspace(0, T, n+1)

# store all paths
paths = np.zeros((num_paths, n+1))

# simulate paths
for i in range(num_paths):

    _, W = brownian_motion(T, n)

    paths[i] = W

# compute average path
mean_path = np.mean(paths, axis=0)

# plot
plt.figure(figsize=(10,6))

# plot individual paths
for i in range(num_paths):

    plt.plot(
        t,
        paths[i],
        color='gray',
        alpha=0.3,
        linewidth=0.8
    )

# plot average path
plt.plot(
    t,
    mean_path,
    color='black',
    linewidth=2,
    label='Average Path'
)

plt.xlabel("t")
plt.ylabel(r"$W_t$")
plt.title("Simulated Brownian Motion Paths")

plt.legend()
plt.grid(True)

plt.show()
```

```python
from scipy.stats import norm

terminal_values = []

for i in range(1000000):

    _, W = brownian_motion(T=1, n=1000)

    terminal_values.append(W[-1])

plt.hist(terminal_values, bins=50, density=True)

x = np.linspace(-5, 5, 1000)

plt.plot(
    x,
    norm.pdf(x, 0, np.sqrt(T)),
    linewidth=2,
    label='Theoretical Density'
)

plt.legend()
plt.show()
```
