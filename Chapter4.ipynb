# Exercise 4.1 — American Put, Call, and Straddle

### Setup
From Figure 1.2.2: $S_0=4, u=2, d=\frac{1}{2}, r=\frac{1}{4}, \tilde{p}=\tilde{q}=\frac{1}{2}$

### (i) American Put: $g_P(s) = (4-s)^+$

| $n$ | $s$ | $v_n(s)$ |
|-----|-----|---------|
| $3$ | $32, 8, 2, 0.5$ | $0, 0, 2, 3.5$ |
| $2$ | $16, 4, 1$ | $0, \frac{4}{5}, 3$ |
| $1$ | $8, 2$ | $\frac{8}{25}, 2$ |
| $0$ | $4$ | $\frac{116}{125}$ |

$$V_0^P = \frac{116}{125}$$

Early exercise at $t=1$: $v_1(2) = 2 = g_P(2)$ → immediate exercise optimal.

### (ii) American Call: $g_C(s) = (s-4)^+$

| $n$ | $s$ | $v_n(s)$ |
|-----|-----|---------|
| $3$ | $32, 8, 2, 0.5$ | $28, 4, 0, 0$ |
| $2$ | $16, 4, 1$ | $\frac{64}{5}, \frac{8}{5}, 0$ |
| $1$ | $8, 2$ | $\frac{144}{25}, \frac{16}{25}$ |
| $0$ | $4$ | $\frac{64}{25}$ |

$$V_0^C = \frac{64}{25}$$

No early exercise optimal (consistent with the fact that American calls on 
non-dividend paying stocks are never optimally exercised early).

### (iii) American Straddle: $g_S(s) = |s-4|$

| $n$ | $s$ | $v_n(s)$ |
|-----|-----|---------|
| $3$ | $32, 8, 2, 0.5$ | $28, 4, 2, 3.5$ |
| $2$ | $16, 4, 1$ | $\frac{64}{5}, \frac{12}{5}, 3$ |
| $1$ | $8, 2$ | $\frac{152}{25}, \frac{54}{25}$ |
| $0$ | $4$ | $\frac{412}{125}$ |

$$V_0^S = \frac{412}{125}$$

### (iv) Why $V_0^S < V_0^P + V_0^C$?

$$V_0^P + V_0^C = \frac{116}{125} + \frac{320}{125} = \frac{436}{125} > \frac{412}{125} = V_0^S$$

**Key insight:** When $s < 4$ and Put is exercised early ($g_P(s) > c_n^P$),
the Call has zero intrinsic value so $V_n^C(s) = c_n^C$:

$$V_n^P(s) + V_n^C(s) = \max\{4-s, c_n^P\} + c_n^C$$

But the Straddle must exercise both simultaneously:

$$V_n^S(s) = \max\{4-s, c_n^P + c_n^C\}$$

When Put is exercised early ($4-s > c_n^P$) but $c_n^P + c_n^C > 4-s$:

$$V_n^P + V_n^C - V_n^S = (4-s) + c_n^C - (c_n^P + c_n^C) = 4-s-c_n^P > 0$$

**The Straddle holder must exercise Put and Call at the same time, 
losing the flexibility to exercise them separately at their respective 
optimal times. This loss of flexibility causes $V_0^S < V_0^P + V_0^C$.**

## Exercise 4.2 — Financing a Loan by Optimal Exercise of American Put

### Setup
American put with strike $K=5$, $S_0=4, u=2, d=\frac{1}{2}, r=\frac{1}{4}$.
Agent borrows $V_0 = 1.36$ at time zero to buy the put.

Loan grows at 25% per period:
- $t=0$: debt $= 1.36$
- $t=1$: debt $= 1.36 \times 1.25 = 1.70$
- $t=2$: debt $= 1.70 \times 1.25 = 2.125$
- $t=3$: debt $= 2.125 \times 1.25 = 2.656$

### Hedging Strategy

$$\Delta_0 = -\frac{13}{90}, \quad \Delta_1(H) = -\frac{1}{12}, \quad \Delta_1(T) = -1$$

### How the Loan is Repaid

**At $t=1$, tail (early exercise optimal):**

Agent exercises put immediately, receiving $g_P(S_1(T)) = g_P(2) = 3$.

$$v_1(T) = 3 \geq 1.70 = \text{loan at } t=1$$

Loan is fully repaid with $3 - 1.70 = 1.30$ remaining.

**At $t=1$, head (continue holding):**

$v_1(H) = 0.48$, loan $= 1.70$.

Agent continues hedging with $\Delta_1(H) = -\frac{1}{12}$ until maturity.

At $t=3$, regardless of outcome:

$$v_3(S_3) = g_P(S_3) \geq 0$$

The put payoff combined with the hedging portfolio covers the loan.

### Key Insight

By the American pricing formula:
$$V_n \geq g_P(S_n) \quad \text{for all } n$$

The put value always exceeds the intrinsic value. By following the 
**optimal exercise strategy** (exercise when $V_n = g_P(S_n)$), the agent 
ensures the put payoff is always sufficient to repay the loan:

$$V_n = v_n(S_n) \geq \text{loan at time } n = \frac{V_0 \cdot (1+r)^n}{1}$$

since $V_0 = 1.36$ is exactly the time-zero value of the put, and the 
discounted put value is a supermartingale under $\tilde{\mathbb{P}}$.
