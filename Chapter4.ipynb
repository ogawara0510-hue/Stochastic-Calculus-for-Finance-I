# Exercise 4.1 — American Put, Call, and Straddle

### Setup
From Figure 1.2.2: $S_0=4, u=2, d=\frac{1}{2}, r=\frac{1}{4}, \tilde{p}=\tilde{q}=\frac{1}{2}$

### (i) American Put: $g_P(s) = (4-s)^+$

| $n$ | $s$ | $v_n(s)$ |
|-----|-----|---------|
| $3$ | $32, 8, 2, 0.5$ | $0, 0, 2, 3.5$ |
| $2$ | $16, 4, 1$ | $0, \frac{4}{5}, 3$ |
| $1$ | $8, 2$ | $\frac{8}{25}, 2$ |
| $0$ | $4$ | $\frac{116}{125}$ |

$$V_0^P = \frac{116}{125}$$

Early exercise at $t=1$: $v_1(2) = 2 = g_P(2)$ → immediate exercise optimal.

### (ii) American Call: $g_C(s) = (s-4)^+$

| $n$ | $s$ | $v_n(s)$ |
|-----|-----|---------|
| $3$ | $32, 8, 2, 0.5$ | $28, 4, 0, 0$ |
| $2$ | $16, 4, 1$ | $\frac{64}{5}, \frac{8}{5}, 0$ |
| $1$ | $8, 2$ | $\frac{144}{25}, \frac{16}{25}$ |
| $0$ | $4$ | $\frac{64}{25}$ |

$$V_0^C = \frac{64}{25}$$

No early exercise optimal (consistent with the fact that American calls on 
non-dividend paying stocks are never optimally exercised early).

### (iii) American Straddle: $g_S(s) = |s-4|$

| $n$ | $s$ | $v_n(s)$ |
|-----|-----|---------|
| $3$ | $32, 8, 2, 0.5$ | $28, 4, 2, 3.5$ |
| $2$ | $16, 4, 1$ | $\frac{64}{5}, \frac{12}{5}, 3$ |
| $1$ | $8, 2$ | $\frac{152}{25}, \frac{54}{25}$ |
| $0$ | $4$ | $\frac{412}{125}$ |

$$V_0^S = \frac{412}{125}$$

### (iv) Why $V_0^S < V_0^P + V_0^C$?

$$V_0^P + V_0^C = \frac{116}{125} + \frac{320}{125} = \frac{436}{125} > \frac{412}{125} = V_0^S$$

**Key insight:** When $s < 4$ and Put is exercised early ($g_P(s) > c_n^P$),
the Call has zero intrinsic value so $V_n^C(s) = c_n^C$:

$$V_n^P(s) + V_n^C(s) = \max\{4-s, c_n^P\} + c_n^C$$

But the Straddle must exercise both simultaneously:

$$V_n^S(s) = \max\{4-s, c_n^P + c_n^C\}$$

When Put is exercised early ($4-s > c_n^P$) but $c_n^P + c_n^C > 4-s$:

$$V_n^P + V_n^C - V_n^S = (4-s) + c_n^C - (c_n^P + c_n^C) = 4-s-c_n^P > 0$$

**The Straddle holder must exercise Put and Call at the same time, 
losing the flexibility to exercise them separately at their respective 
optimal times. This loss of flexibility causes $V_0^S < V_0^P + V_0^C$.**

## Exercise 4.2 — Financing a Loan by Optimal Exercise of American Put

### Setup
American put with strike $K=5$, $S_0=4, u=2, d=\frac{1}{2}, r=\frac{1}{4}$.
Agent borrows $V_0 = 1.36$ at time zero to buy the put.

Loan grows at 25% per period:
- $t=0$: debt $= 1.36$
- $t=1$: debt $= 1.36 \times 1.25 = 1.70$
- $t=2$: debt $= 1.70 \times 1.25 = 2.125$
- $t=3$: debt $= 2.125 \times 1.25 = 2.656$

### Hedging Strategy

$$\Delta_0 = -\frac{13}{90}, \quad \Delta_1(H) = -\frac{1}{12}, \quad \Delta_1(T) = -1$$

### How the Loan is Repaid

**At $t=1$, tail (early exercise optimal):**

Agent exercises put immediately, receiving $g_P(S_1(T)) = g_P(2) = 3$.

$$v_1(T) = 3 \geq 1.70 = \text{loan at } t=1$$

Loan is fully repaid with $3 - 1.70 = 1.30$ remaining.

**At $t=1$, head (continue holding):**

$v_1(H) = 0.48$, loan $= 1.70$.

Agent continues hedging with $\Delta_1(H) = -\frac{1}{12}$ until maturity.

At $t=3$, regardless of outcome:

$$v_3(S_3) = g_P(S_3) \geq 0$$

The put payoff combined with the hedging portfolio covers the loan.

### Key Insight

By the American pricing formula:
$$V_n \geq g_P(S_n) \quad \text{for all } n$$

The put value always exceeds the intrinsic value. By following the 
**optimal exercise strategy** (exercise when $V_n = g_P(S_n)$), the agent 
ensures the put payoff is always sufficient to repay the loan:

$$V_n = v_n(S_n) \geq \text{loan at time } n = \frac{V_0 \cdot (1+r)^n}{1}$$

since $V_0 = 1.36$ is exactly the time-zero value of the put, and the 
discounted put value is a supermartingale under $\tilde{\mathbb{P}}$.

# Exercise 4.3 — Path-Dependent American Derivative (Asian Put)

### Setup
From Figure 1.2.2: $S_0=4, u=2, d=\frac{1}{2}, r=\frac{1}{4}, \tilde{p}=\tilde{q}=\frac{1}{2}$

Intrinsic value at time $n$:
$$g_n = \left(4 - \frac{1}{n+1}\sum_{j=0}^{n} S_j\right)^+$$

This is a put on the **average stock price** from time 0 to time $n$.

### State Space
Since the intrinsic value depends on the path, the state is $(S_n, Y_n)$ 
where $Y_n = \sum_{j=0}^{n} S_j$.

### Terminal Payoffs ($t=3$)

| Path | $S_3$ | $Y_3$ | $g_3 = (4 - Y_3/4)^+$ |
|------|-------|-------|----------------------|
| $HHH$ | $32$ | $60$ | $0$ |
| $HHT$ | $8$  | $36$ | $0$ |
| $HTH$ | $8$  | $24$ | $0$ |
| $THH$ | $8$  | $18$ | $0$ |
| $HTT$ | $2$  | $18$ | $0$ |
| $THT$ | $2$  | $12$ | $1$ |
| $TTH$ | $2$  | $9$  | $1.75$ |
| $TTT$ | $0.5$ | $7.5$ | $2.125$ |

### Option Values

**$t=2$:**

| Node | $S_2$ | $Y_2$ | $g_2$ | continuation | $V_2$ |
|------|-------|-------|-------|-------------|-------|
| $HH$ | $16$ | $28$ | $0$ | $0$ | $0$ |
| $HT$ | $4$  | $20$ | $0$ | $0$ | $0$ |
| $TH$ | $4$  | $14$ | $\frac{1}{3}$ | $\frac{2}{5}$ | $\frac{2}{5}$ |
| $TT$ | $1$  | $7$  | $\frac{5}{3}$ | $\frac{59}{40}$ | $\frac{5}{3}$ |

**$t=1$:**

| Node | $S_1$ | $Y_1$ | $g_1$ | continuation | $V_1$ |
|------|-------|-------|-------|-------------|-------|
| $H$ | $8$ | $12$ | $0$ | $0$ | $0$ |
| $T$ | $2$ | $6$  | $1$ | $\frac{4}{5} \cdot \frac{1}{2}(\frac{2}{5}+\frac{5}{3})$ | $1$ |

**$t=0$:**

$$V_0 = \max\left\{0, \frac{4}{5}\left[\frac{1}{2} \cdot 0 + \frac{1}{2} \cdot 1\right]\right\} = \frac{2}{5}$$

### Optimal Exercise Policy

- $t=1$, $T$: exercise immediately ($V_1(T) = g_1(T) = 1$)
- $t=2$, $TT$: exercise immediately ($V_2(TT) = g_2(TT) = \frac{5}{3}$)
- All other nodes: hold until maturity

```python
from itertools import product
from functools import lru_cache

# Parameters
S0, u, d, r = 4, 2, 0.5, 0.25
p_tilde, q_tilde = 0.5, 0.5

def intrinsic_value(n, Y_n):
    """g_n = (4 - Y_n/(n+1))^+"""
    return max(4 - Y_n / (n + 1), 0)

def american_asian_put(S0, u, d, r, m):
    """
    American derivative with path-dependent intrinsic value.
    State: (n, S_n, Y_n) where Y_n = sum of S_0,...,S_n
    """
    @lru_cache(maxsize=None)
    def V(n, S, Y):
        g = intrinsic_value(n, Y)
        if n == m:
            return g
        
        S_up   = S * u
        S_down = S * d
        Y_up   = Y + S_up
        Y_down = Y + S_down
        
        continuation = (p_tilde * V(n+1, S_up, Y_up) + 
                       q_tilde * V(n+1, S_down, Y_down)) / (1+r)
        
        return max(g, continuation)
    
    return V

# Compute
V = american_asian_put(S0, u, d, r, m=3)

print(f"V0(4, 4) = {V(0, S0, S0):.6f}")
print()

# t=1
print(f"V1(H): S=8, Y=12 → {V(1, 8, 12):.6f}")
print(f"V1(T): S=2, Y=6  → {V(1, 2, 6):.6f}")
print()

# t=2
print(f"V2(HH): S=16, Y=28 → {V(2, 16, 28):.6f}")
print(f"V2(HT): S=4,  Y=20 → {V(2, 4,  20):.6f}")
print(f"V2(TH): S=4,  Y=14 → {V(2, 4,  14):.6f}")
print(f"V2(TT): S=1,  Y=7  → {V(2, 1,  7):.6f}")
print()

# t=3 (payoffs)
print("t=3 payoffs:")
paths = {
    'HHH': (32, 60), 'HHT': (8, 36), 'HTH': (8, 24), 'HTT': (2, 18),
    'THH': (8, 18),  'THT': (2, 12), 'TTH': (2, 9),  'TTT': (0.5, 7.5)
}
for path, (S, Y) in paths.items():
    print(f"  {path}: S={S}, Y={Y} → g={intrinsic_value(3, Y):.4f}")
```

## Exercise 4.5 — Stopping Times and Optimal Exercise

### Setup
American put with strike $K=5$, $S_0=4, u=2, d=\frac{1}{2}, r=\frac{1}{4}, \tilde{p}=\tilde{q}=\frac{1}{2}$

### All 26 Stopping Times in $S_0$

**Structure:** 5 choices on the H-side × 5 choices on the T-side + 1 choice ($\tau=0$) = 26 stopping times.

**H-side (5 choices):**
1. $\tau(H\cdots) = 1$ (exercise at $t=1$ on all H-paths)
2. $\tau(HH\cdots) = 2, \tau(HT\cdots) = 2$
3. $\tau(HH\cdots) = 2, \tau(HTH) = 3, \tau(HTT) = 3$
4. $\tau(HHH) = 3, \tau(HHT) = 3, \tau(HT\cdots) = 2$
5. $\tau(HHH) = 3, \tau(HHT) = 3, \tau(HTH) = 3, \tau(HTT) = 3$

**T-side (5 choices):**
1. $\tau(T\cdots) = 1$
2. $\tau(TH\cdots) = 2, \tau(TT\cdots) = 2$
3. $\tau(TH\cdots) = 2, \tau(TTH) = 3, \tau(TTT) = 3$
4. $\tau(THH) = 3, \tau(THT) = 3, \tau(TT\cdots) = 2$
5. $\tau(THH) = 3, \tau(THT) = 3, \tau(TTH) = 3, \tau(TTT) = 3$

**Total:** $5 \times 5 + 1 = 26$ stopping times $\checkmark$

The put is out of the money when $S_n > 5$:
$$S_1(H)=8, \quad S_2(HH)=16, \quad S_3(HHH)=32, \quad S_3(HHT)=8$$

These nodes must never be exercised, so on the H-side:
$$\tau(HHH)=3, \quad \tau(HHT)=3 \quad \text{(fixed)}$$

This leaves only 2 choices on the H-side, while the T-side remains unconstrained (5 choices):

$$\text{Total: } 2 \times 5 + 1\ (\tau=0) = 11 \text{ stopping times} \checkmark$$

The optimal stopping time $\tau^*$ is:
$$\tau^*(HHH) = 3, \quad \tau^*(HHT) = 3, \quad \tau^*(HT\cdots) = 2, \quad \tau^*(T\cdots) = 1$$

| Path | $\tilde{\mathbb{P}}$ | $\tau^*$ | $G_{\tau^*}$ | $\left(\frac{4}{5}\right)^{\tau^*} G_{\tau^*}$ |
|------|---------------------|---------|-------------|------------------------------------------------|
| $HHH$ | $\frac{1}{8}$ | $3$ | $0$ | $0$ |

| $THT$ | $\frac{1}{8}$ | $1$ | $3$ | $\frac{12}{5}$ |

$$\tilde{\mathbb{E}}\left[\left(\frac{4}{5}\right)^{\tau^*} G_{\tau^*}\right] 
= \frac{1}{8}\left[0 + 0 + \frac{16}{25} + \frac{16}{25} + 4 \times \frac{12}{5}\right]
= \frac{1}{8} \cdot \frac{272}{25} = \frac{34}{25} = 1.36 \checkmark$$

### Key Insight
Among all 11 stopping times that never exercise out of the money, 
$\tau^*$ achieves the maximum expected discounted payoff of $1.36$, 
confirming that $\tau^*$ is the optimal stopping time and that the 
American put price $V_0 = 1.36$ is correctly computed. $\blacksquare$

# Exercise 4.6 — Estimating American Put Prices

### Setup
For each $n$, let $G_n$ be a random variable depending on the first $n$ 
coin tosses. The time-zero value of a derivative that must be exercised 
by time $N$ is:

$$V_0 = \max_{\tau \in S_0, \tau \leq N} \tilde{\mathbb{E}}\left[\frac{G_\tau}{(1+r)^\tau}\right]$$

### (i) Optimal Exercise Policy for $G_n = K - S_n$

$$\tilde{\mathbb{E}}\left[\frac{K - S_\tau}{(1+r)^\tau}\right] 
= \tilde{\mathbb{E}}\left[\frac{K}{(1+r)^\tau}\right] - \tilde{\mathbb{E}}\left[\frac{S_\tau}{(1+r)^\tau}\right]$$

By the Optional Sampling Theorem, since the discounted stock price is a 
martingale under $\tilde{\mathbb{P}}$:

$$\tilde{\mathbb{E}}\left[\frac{S_\tau}{(1+r)^\tau}\right] = S_0$$

Since $\frac{K}{(1+r)^\tau}$ is maximized by choosing $\tau = 0$:

$$V_0 = K - S_0 \quad \blacksquare$$

The optimal policy is to **exercise immediately at time zero**.

### (ii) Upper Bound: $V_0^{AP} \leq K - S_0 + V_0^{EC}$

For any stopping time $\tau$ and any outcome:

$$\underbrace{(K - S_\tau)^+}_{\text{American put payoff}} 
\leq \underbrace{(K - S_\tau)}_{\text{derivative}} + \underbrace{(S_N - K)^+}_{\text{European call}}$$

since $(K-S_\tau)^+ \leq (K-S_\tau) + (S_N-K)^+$ holds in all cases:

- $S_\tau \leq K$: $(K-S_\tau) \leq (K-S_\tau) + (S_N-K)^+$ ✅
- $S_\tau > K$: $0 \leq (K-S_\tau) + (S_N-K)^+$ ✅ (European call compensates)

Taking risk-neutral expectations and maximizing over $\tau$:

$$V_0^{AP} = \max_\tau \tilde{\mathbb{E}}\left[\frac{(K-S_\tau)^+}{(1+r)^\tau}\right] 
\leq \max_\tau \tilde{\mathbb{E}}\left[\frac{(K-S_\tau) + (S_N-K)^+}{(1+r)^\tau}\right]$$

$$\leq (K - S_0) + V_0^{EC} \quad \blacksquare$$

### (iii) Lower Bound: $V_0^{AP} \geq \frac{K}{(1+r)^N} - S_0 + V_0^{EC}$

By Put-Call Parity (Exercise 2.11):

$$V_0^{EC} - V_0^{EP} = S_0 - \frac{K}{(1+r)^N}$$

$$\implies V_0^{EP} = V_0^{EC} - S_0 + \frac{K}{(1+r)^N}$$

Since American put $\geq$ European put ($V_0^{AP} \geq V_0^{EP}$):

$$V_0^{AP} \geq V_0^{EC} - S_0 + \frac{K}{(1+r)^N} \quad \blacksquare$$

### Summary: Price Bounds for American Put

$$\underbrace{V_0^{EC} - S_0 + \frac{K}{(1+r)^N}}_{\text{lower bound}} 
\leq V_0^{AP} \leq 
\underbrace{V_0^{EC} - S_0 + K}_{\text{upper bound}}$$

The gap between the bounds is:

$$K - \frac{K}{(1+r)^N} = K\left(1 - \frac{1}{(1+r)^N}\right) > 0$$

This gap represents the **early exercise premium** — the additional value 
from being able to exercise before maturity. As $N \to \infty$ or $r \to 0$, 
the bounds converge and early exercise becomes less valuable.

# Exercise 4.7 — Optimal Exercise for $G_n = S_n - K$

### Problem
A derivative security permits its owner to buy one share of stock for 
payment $K$ at any time up to expiration $N$. If not exercised by time $N$, 
it must be exercised then.

$$V_0 = \max_{\tau \in S_0, \tau \leq N} \tilde{\mathbb{E}}\left[\frac{S_\tau - K}{(1+r)^\tau}\right]$$

### Solution

$$\tilde{\mathbb{E}}\left[\frac{S_\tau - K}{(1+r)^\tau}\right] 
= \tilde{\mathbb{E}}\left[\frac{S_\tau}{(1+r)^\tau}\right] - \tilde{\mathbb{E}}\left[\frac{K}{(1+r)^\tau}\right]$$

By the Optional Sampling Theorem, since the discounted stock price is a 
martingale under $\tilde{\mathbb{P}}$:

$$\tilde{\mathbb{E}}\left[\frac{S_\tau}{(1+r)^\tau}\right] = S_0$$

Since $\frac{K}{(1+r)^\tau}$ is **decreasing** in $\tau$, it is minimized 
by choosing $\tau = N$:

$$V_0 = S_0 - \frac{K}{(1+r)^N} \quad \blacksquare$$

The optimal policy is to **delay exercise until time $N$**.

### Comparison with Exercise 4.6(i)

| Derivative | $G_n$ | Optimal $\tau$ | $V_0$ |
|-----------|-------|---------------|-------|
| Ex 4.6(i) — sell stock | $K - S_n$ | $\tau = 0$ | $K - S_0$ |
| Ex 4.7 — buy stock | $S_n - K$ | $\tau = N$ | $S_0 - \frac{K}{(1+r)^N}$ |

### Key Insight

- **Selling:** Exercise immediately at $\tau=0$ — receive payment $K$ as 
early as possible to maximize its present value.
- **Buying:** Delay exercise until $\tau=N$ — defer payment $K$ as long 
as possible to minimize its present value.

Note that $V_0 = S_0 - \frac{K}{(1+r)^N}$ is exactly the **forward 
contract price** $F_0$ from Exercise 2.11(iii), confirming that when 
forced to buy at expiration, the fair price equals the forward price.
