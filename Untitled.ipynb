# Exercise 6.1 — Fundamental Properties of Conditional Expectations

### Setup
Using Definition 6.2.2, for a random variable $X$:

$$\tilde{\mathbb{E}}_n[X](\bar{\omega}_1\ldots\bar{\omega}_n) = \sum_{\bar{\omega}_{n+1},\ldots,\bar{\omega}_N} X(\bar{\omega}_1\ldots\bar{\omega}_N) \cdot \tilde{\mathbb{P}}\{\omega_{n+1}=\bar{\omega}_{n+1},\ldots,\omega_N=\bar{\omega}_N|\omega_1=\bar{\omega}_1,\ldots,\omega_n=\bar{\omega}_n\}$$

### (i) Linearity

$$\tilde{\mathbb{E}}_n[c_1X + c_2Y](\bar{\omega}_1\ldots\bar{\omega}_n)$$

$$= \sum_{\bar{\omega}_{n+1},\ldots,\bar{\omega}_N} (c_1X + c_2Y)(\bar{\omega}_1\ldots\bar{\omega}_N) \cdot \tilde{\mathbb{P}}\{\cdots\}$$

$$= c_1\sum_{\bar{\omega}_{n+1},\ldots,\bar{\omega}_N} X(\bar{\omega}_1\ldots\bar{\omega}_N) \cdot \tilde{\mathbb{P}}\{\cdots\} + c_2\sum_{\bar{\omega}_{n+1},\ldots,\bar{\omega}_N} Y(\bar{\omega}_1\ldots\bar{\omega}_N) \cdot \tilde{\mathbb{P}}\{\cdots\}$$

$$= c_1\tilde{\mathbb{E}}_n[X] + c_2\tilde{\mathbb{E}}_n[Y] \quad \blacksquare$$

### (ii) Taking Out What is Known

If $X$ depends only on the first $n$ coin tosses:

$$\tilde{\mathbb{E}}_n[XY](\bar{\omega}_1\ldots\bar{\omega}_n) = \sum_{\bar{\omega}_{n+1},\ldots,\bar{\omega}_N} X(\bar{\omega}_1\ldots\bar{\omega}_n) \cdot Y(\bar{\omega}_1\ldots\bar{\omega}_N) \cdot \tilde{\mathbb{P}}\{\cdots\}$$

Since $X(\bar{\omega}_1\ldots\bar{\omega}_n)$ does not depend on $\bar{\omega}_{n+1},\ldots,\bar{\omega}_N$:

$$= X(\bar{\omega}_1\ldots\bar{\omega}_n) \sum_{\bar{\omega}_{n+1},\ldots,\bar{\omega}_N} Y(\bar{\omega}_1\ldots\bar{\omega}_N) \cdot \tilde{\mathbb{P}}\{\cdots\} = X \cdot \tilde{\mathbb{E}}_n[Y] \quad \blacksquare$$

### (iii) Iterated Conditioning

$$\tilde{\mathbb{E}}_n[\tilde{\mathbb{E}}_m[X]](\bar{\omega}_1\ldots\bar{\omega}_n)$$

$$= \sum_{\bar{\omega}_{n+1},\ldots,\bar{\omega}_m} \sum_{\bar{\omega}_{m+1},\ldots,\bar{\omega}_N} X(\bar{\omega}_1\ldots\bar{\omega}_N) \cdot \tilde{\mathbb{P}}\{\omega_{m+1}=\bar{\omega}_{m+1},\ldots|\omega_1=\bar{\omega}_1,\ldots,\omega_m=\bar{\omega}_m\} \cdot \tilde{\mathbb{P}}\{\omega_{n+1}=\bar{\omega}_{n+1},\ldots,\omega_m=\bar{\omega}_m|\omega_1=\bar{\omega}_1,\ldots,\omega_n=\bar{\omega}_n\}$$

By the multiplication rule for conditional probabilities:

$$\tilde{\mathbb{P}}\{\cdots|\omega_1,\ldots,\omega_m\} \cdot \tilde{\mathbb{P}}\{\cdots|\omega_1,\ldots,\omega_n\} = \tilde{\mathbb{P}}\{\omega_{n+1}=\bar{\omega}_{n+1},\ldots,\omega_N=\bar{\omega}_N|\omega_1=\bar{\omega}_1,\ldots,\omega_n=\bar{\omega}_n\}$$

Therefore:

$$= \sum_{\bar{\omega}_{n+1},\ldots,\bar{\omega}_N} X(\bar{\omega}_1\ldots\bar{\omega}_N) \cdot \tilde{\mathbb{P}}\{\omega_{n+1}=\bar{\omega}_{n+1},\ldots,\omega_N=\bar{\omega}_N|\omega_1=\bar{\omega}_1,\ldots,\omega_n=\bar{\omega}_n\} = \tilde{\mathbb{E}}_n[X] \quad \blacksquare$$

### (iv) Independence

If $X$ depends only on coin tosses $n+1$ through $N$:

$$X(\bar{\omega}_1\ldots\bar{\omega}_N) = X(\bar{\omega}_{n+1}\ldots\bar{\omega}_N)$$

and the conditional probability equals the unconditional probability:

$$\tilde{\mathbb{P}}\{\omega_{n+1}=\bar{\omega}_{n+1},\ldots,\omega_N=\bar{\omega}_N|\omega_1=\bar{\omega}_1,\ldots,\omega_n=\bar{\omega}_n\} = \tilde{\mathbb{P}}\{\omega_{n+1}=\bar{\omega}_{n+1},\ldots,\omega_N=\bar{\omega}_N\}$$

Therefore:

$$\tilde{\mathbb{E}}_n[X](\bar{\omega}_1\ldots\bar{\omega}_n) = \sum_{\bar{\omega}_{n+1},\ldots,\bar{\omega}_N} X(\bar{\omega}_{n+1}\ldots\bar{\omega}_N) \cdot \tilde{\mathbb{P}}\{\omega_{n+1}=\bar{\omega}_{n+1},\ldots,\omega_N=\bar{\omega}_N\} = \tilde{\mathbb{E}}[X] \quad \blacksquare$$

### (v) Conditional Jensen's Inequality

Since $\varphi$ is convex, the tangent line inequality holds:

$$\varphi(x) \geq \varphi(y) + \varphi'(y)(x-y)$$

Setting $y = \tilde{\mathbb{E}}_n[X]$ and $x = X$, then taking $\tilde{\mathbb{E}}_n$:

$$\tilde{\mathbb{E}}_n[\varphi(X)] \geq \varphi(\tilde{\mathbb{E}}_n[X]) + \varphi'(\tilde{\mathbb{E}}_n[X]) \cdot \tilde{\mathbb{E}}_n[X - \tilde{\mathbb{E}}_n[X]]$$

Since $\varphi'(\tilde{\mathbb{E}}_n[X])$ is known at time $n$ (by (ii)):

$$\tilde{\mathbb{E}}_n[X - \tilde{\mathbb{E}}_n[X]] = \tilde{\mathbb{E}}_n[X] - \tilde{\mathbb{E}}_n[X] = 0$$

Therefore:

$$\tilde{\mathbb{E}}_n[\varphi(X)] \geq \varphi(\tilde{\mathbb{E}}_n[X]) \quad \blacksquare$$

# Exercise 6.2 — Discounted Static Hedge Portfolio is a Martingale

### Setup
From the proof of Theorem 6.3.2, at time $n$ an agent:
1. Sells the forward contract (zero cost)
2. Shorts $\frac{S_n}{B_{n,m}}$ units of $m$-maturity zero-coupon bonds → receives $S_n$
3. Uses $S_n$ to buy one unit of the asset

### Portfolio Value at Time $k$ ($n \leq k \leq m$)

$$X_k = \underbrace{S_k}_{\text{long asset}} - \underbrace{\frac{S_n}{B_{n,m}} \cdot B_{k,m}}_{\text{short bond position}}$$

### Discounted Portfolio Value

$$\frac{X_k}{D_k} = \frac{S_k}{D_k} - \frac{S_n}{B_{n,m}} \cdot \frac{B_{k,m}}{D_k}$$

### Verification: Martingale Property

$$\tilde{\mathbb{E}}_k\left[\frac{X_{k+1}}{D_{k+1}}\right] 
= \tilde{\mathbb{E}}_k\left[\frac{S_{k+1}}{D_{k+1}}\right] - \frac{S_n}{B_{n,m}}\tilde{\mathbb{E}}_k\left[\frac{B_{k+1,m}}{D_{k+1}}\right]$$

By Theorem 6.2.6:
- Discounted asset price is a martingale: $\tilde{\mathbb{E}}_k\left[\frac{S_{k+1}}{D_{k+1}}\right] = \frac{S_k}{D_k}$
- Discounted bond price is a martingale: $\tilde{\mathbb{E}}_k\left[\frac{B_{k+1,m}}{D_{k+1}}\right] = \frac{B_{k,m}}{D_k}$

Therefore:

$$\tilde{\mathbb{E}}_k\left[\frac{X_{k+1}}{D_{k+1}}\right] 
= \frac{S_k}{D_k} - \frac{S_n}{B_{n,m}} \cdot \frac{B_{k,m}}{D_k} 
= \frac{X_k}{D_k} \quad \blacksquare$$

### Key Insight
The static hedge portfolio requires **no rebalancing** between time $n$ and $m$,
yet its discounted value is still a martingale. This confirms that the
forward price $\text{For}_{n,m} = \frac{S_n}{B_{n,m}}$ is the unique
no-arbitrage price, since any deviation would create a non-zero initial
portfolio value while maintaining a martingale — a contradiction.

# Exercise 6.3 — Risk-Neutral Pricing of Interest Rate Contract

### Problem
Show that the risk-neutral price at time $n$ of a contract paying $R_m$ 
at time $m+1$ equals $B_{n,m} - B_{n,m+1}$:

$$\frac{1}{D_n}\tilde{\mathbb{E}}_n[D_{m+1}R_m] = B_{n,m} - B_{n,m+1}$$

### Proof

Using $R_m = \frac{1}{B_{m,m+1}} - 1$:

$$\frac{1}{D_n}\tilde{\mathbb{E}}_n[D_{m+1}R_m] 
= \frac{1}{D_n}\tilde{\mathbb{E}}_n\left[\frac{D_{m+1}}{B_{m,m+1}}\right] 
- \frac{1}{D_n}\tilde{\mathbb{E}}_n[D_{m+1}]$$

**Second term:**

By definition of $B_{n,m+1}$:

$$\frac{1}{D_n}\tilde{\mathbb{E}}_n[D_{m+1}] = B_{n,m+1}$$

**First term:**

By Tower Property:

$$\frac{1}{D_n}\tilde{\mathbb{E}}_n\left[\frac{D_{m+1}}{B_{m,m+1}}\right] 
= \frac{1}{D_n}\tilde{\mathbb{E}}_n\left[\tilde{\mathbb{E}}_m\left[\frac{D_{m+1}}{B_{m,m+1}}\right]\right]$$

Since $\frac{1}{B_{m,m+1}}$ is known at time $m$ (taking out what is known):

$$= \frac{1}{D_n}\tilde{\mathbb{E}}_n\left[\frac{1}{B_{m,m+1}}\tilde{\mathbb{E}}_m[D_{m+1}]\right]$$

Since $\tilde{\mathbb{E}}_m[D_{m+1}] = D_m \cdot B_{m,m+1}$:

$$= \frac{1}{D_n}\tilde{\mathbb{E}}_n\left[\frac{D_m \cdot B_{m,m+1}}{B_{m,m+1}}\right] 
= \frac{1}{D_n}\tilde{\mathbb{E}}_n[D_m] = B_{n,m}$$

**Combining:**

$$\frac{1}{D_n}\tilde{\mathbb{E}}_n[D_{m+1}R_m] = B_{n,m} - B_{n,m+1} \quad \blacksquare$$

### Key Insight
This confirms that Theorem 6.3.5 is consistent with the general 
risk-neutral pricing formula. The price $B_{n,m} - B_{n,m+1}$ can be 
derived either by:
1. **Static hedging** (Theorem 6.3.5) — constructing a replicating portfolio
2. **Risk-neutral pricing** (this exercise) — taking discounted expectation

Both methods give the same answer, confirming no-arbitrage consistency.

# Exercise 6.4 — Hedging a Short Position in a Caplet

### Setup
From Example 6.3.9: Caplet paying $(R_2 - \frac{1}{3})^+$ at time three.

**Payoffs:**
$$V_3(HH) = \frac{2}{3}, \quad V_3(HT) = V_3(TH) = V_3(TT) = 0$$

**Time-two prices (by discounting):**
$$V_2(HH) = \frac{1}{1+R_2(HH)}V_3(HH) = \frac{1}{2} \cdot \frac{2}{3} = \frac{1}{3}, \quad V_2(HT) = V_2(TH) = V_2(TT) = 0$$

**Time-zero price:** $V_0 = \frac{2}{21}$ (from Example 6.3.9)

### (i) Time-one Prices

**$V_1(H)$:** Using $R_1(H) = \frac{1}{6}$, risk-neutral probabilities $\tilde{p}=\frac{2}{3}, \tilde{q}=\frac{1}{3}$:

$$V_1(H) = \frac{1}{1+\frac{1}{6}}\left[\frac{2}{3} \cdot \frac{1}{3} + \frac{1}{3} \cdot 0\right] = \frac{6}{7} \cdot \frac{2}{9} = \frac{4}{21}$$

**$V_1(T)$:** Using $R_1(T) = \frac{2}{5}$, $V_2(TH) = V_2(TT) = 0$:

$$V_1(T) = \frac{1}{1+\frac{2}{5}}\left[\frac{1}{2} \cdot 0 + \frac{1}{2} \cdot 0\right] = 0$$

### (ii) Hedging from Time Zero to Time One

Portfolio: $\alpha_0$ in money market, $\delta_0$ units of maturity-two bond.

**Bond prices at time one:**
$$B_{1,2}(H) = \frac{1}{1+R_1(H)} = \frac{6}{7}, \quad B_{1,2}(T) = \frac{1}{1+R_1(T)} = \frac{5}{7}$$

**System of equations:**
$$\alpha_0 + \delta_0 \cdot \frac{6}{7} = \frac{4}{21} \quad (H)$$
$$\alpha_0 + \delta_0 \cdot \frac{5}{7} = 0 \quad (T)$$

**Solution:**
$$\delta_0 = \frac{4}{3}, \quad \alpha_0 = -\frac{20}{21}$$

**Verification:**
$$\alpha_0 + \delta_0 \cdot B_{0,2} = -\frac{20}{21} + \frac{4}{3} \cdot \frac{11}{14} = \frac{2}{21} \checkmark$$

**Why maturity-two bond?**

$V_1$ depends on $R_1$, which determines $B_{1,2}$. The maturity-two bond
matures at time two, so its time-one value $B_{1,2}$ varies with $R_1$,
providing the necessary sensitivity to hedge. The maturity-three bond's
time-one value $B_{1,3} = \frac{4}{7}$ is the same for both $H$ and $T$,
so it cannot distinguish between the two outcomes.

### (iii) Hedging from Time One to Time Two

**Bond prices at time two:**
$$B_{2,3}(HH) = \frac{1}{2}, \quad B_{2,3}(HT) = 1, \quad B_{2,3}(TH) = B_{2,3}(TT) = \frac{4}{5}$$

**H-side:** Portfolio $\alpha_1(H)$ in money market, $\delta_1(H)$ units of maturity-three bond.

System of equations:
$$\alpha_1(H) \cdot \frac{7}{6} + \delta_1(H) \cdot \frac{1}{2} = \frac{1}{3} \quad (HH)$$
$$\alpha_1(H) \cdot \frac{7}{6} + \delta_1(H) \cdot 1 = 0 \quad (HT)$$

Solution:
$$\delta_1(H) = -\frac{2}{3}, \quad \alpha_1(H) = \frac{4}{7}$$

Verification:
$$\alpha_1(H) + \delta_1(H) \cdot B_{1,3}(H) = \frac{4}{7} - \frac{2}{3} \cdot \frac{4}{7} = \frac{4}{21} = X_1(H) \checkmark$$

**T-side:** Since $V_2(TH) = V_2(TT) = 0$:
$$\alpha_1(T) = 0, \quad \delta_1(T) = 0$$

**Why maturity-three bond?**

At time two, $V_2$ depends on $R_2$, which determines $B_{2,3}$.
The maturity-three bond's time-two value $B_{2,3}$ varies with $R_2$,
providing the necessary sensitivity. The maturity-two bond has
$B_{2,2} = 1$ regardless of $R_2$, so it cannot hedge against
$R_2$ uncertainty.
